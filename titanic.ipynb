```python
# ============================================================
# WEEK 1 - ML FUNDAMENTALS + DATA PREPROCESSING
# TITANIC DATASET
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD TITANIC DATASET
# ============================================================

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("\nDataset loaded successfully!")


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFirst 5 rows of the dataset:")
display(df.head())


# ============================================================
# 4. DISPLAY LAST 5 ROWS
# ============================================================

print("\nLast 5 rows of the dataset:")
display(df.tail())


# ============================================================
# 5. DATASET SHAPE
# ============================================================

print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# 6. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# 7. STATISTICAL SUMMARY
# ============================================================

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# 8. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())


# ============================================================
# 9. HANDLE MISSING VALUES
# ============================================================

# Fill Age missing values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill Fare missing values with median
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

# Fill Embarked missing values with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print("\nMissing values handled successfully!")


# ============================================================
# 10. CHECK MISSING VALUES AFTER CLEANING
# ============================================================

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())


# ============================================================
# 11. LABEL ENCODING
# ============================================================

label_encoder = LabelEncoder()

df['Sex_Label'] = label_encoder.fit_transform(df['Sex'])

print("\nLabel Encoding of Sex:")
display(df[['Sex', 'Sex_Label']].head(10))


# ============================================================
# 12. ONE-HOT ENCODING
# ============================================================

embarked_encoded = pd.get_dummies(
    df['Embarked'],
    prefix='Embarked',
    dtype=int
)

df = pd.concat([df, embarked_encoded], axis=1)

print("\nOne-Hot Encoding of Embarked:")
display(df[['Embarked', 'Embarked_C', 'Embarked_Q', 'Embarked_S']].head(10))


# ============================================================
# 13. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df['Age'],
    bins=30,
    kde=True
)

plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()


# ============================================================
# 14. SURVIVAL COUNT
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Survived'
)

plt.title('Titanic Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')

plt.show()


# ============================================================
# 15. SURVIVAL BY SEX
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Sex',
    hue='Survived'
)

plt.title('Survival by Sex')
plt.xlabel('Sex')
plt.ylabel('Number of Passengers')

plt.show()


# ============================================================
# 16. SURVIVAL BY PASSENGER CLASS
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Pclass',
    hue='Survived'
)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()


# ============================================================
# 17. FINAL CLEANED DATASET
# ============================================================

print("\nFinal Cleaned Dataset:")
display(df.head(10))


# ============================================================
# 18. FINAL DATASET INFORMATION
# ============================================================

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Dataset Columns:")
print(df.columns.tolist())


# ============================================================
# 19. FINAL MISSING VALUE CHECK
# ============================================================

print("\nFinal Missing Value Check:")
print(df.isnull().sum())


# ============================================================
# 20. TRAIN-TEST SPLIT
# ============================================================

# Select simple numerical features for demonstration
features = [
    'Pclass',
    'Age',
    'SibSp',
    'Parch',
    'Fare',
    'Sex_Label',
    'Embarked_C',
    'Embarked_Q',
    'Embarked_S'
]

X = df[features]
y = df['Survived']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTrain-Test Split:")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)


# ============================================================
# 21. SAVE CLEANED DATASET
# ============================================================

output_file = "titanic_cleaned.csv"

df.to_csv(
    output_file,
    index=False
)

print("\nCleaned dataset saved as:", output_file)


# ============================================================
# 22. DOWNLOAD CSV FROM GOOGLE COLAB
# ============================================================

from google.colab import files

files.download(output_file)

print("\n================================================")
print("WEEK 1 ASSIGNMENT COMPLETED SUCCESSFULLY!")
print("================================================")
```
